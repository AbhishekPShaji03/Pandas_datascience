## DATA CLEANING (REAL-WORLD SCENARIOS)

7.Identify missing values in each column

```python
print(df.isnull().sum())
```

8.Remove rows where "Rating" is missing

```python
df = df.dropna(subset=["Rating"])
```

9.Fill missing values in "Type" with the most frequent value (mode)

```python
df["Type"] = df["Type"].fillna(df["Type"].mode()[0])
```

10.Remove duplicate records from the dataset

```python
df = df.drop_duplicates()
```

11.Clean "Installs" (remove '+', ',' and non-numeric entries if present, convert to int)

```python
df["Installs"] = df["Installs"].astype(int)
print(df["Installs"].head())
```

12.Clean "Price" (remove '$' and convert to float)

```python
df = df[df["Price"] != "Everyone"]
df["Price"] = df["Price"].astype(float)
print(df["Price"].head())
```

13.Convert "Reviews" column to integer type

```python
df["Reviews"] = df["Reviews"].fillna(0).astype(int)
```

14.Convert "Last Updated" column to datetime format

```python
df["Last Updated"] = pd.to_datetime(df["Last Updated"])
print(df["Last Updated"].head())
```

## 3. STRING OPERATIONS

15.Convert all app names to lowercase

```python
df["App"] = df["App"].str.lower()
print(df["App"].head())
```

16.Convert category names to uppercase

```python
df["Category"] = df["Category"].str.upper()
print(df["Category"].head())
```

17.Split the "Genres" column into multiple parts (takes the first genre)

```python
df["Primary_Genre"] = df["Genres"].str.split(";").str[0]
print(df[["Primary_Genre"]].head())
```

18.Replace unwanted text patterns in "Category" (e.g., replacing underscores with spaces)

```python
df["Category"] = df["Category"].str.replace("_", " ")
df.Category
```

### 4. FILTERING & SELECTION

19.Filter apps with rating greater than 4.5

```python
highly_rating = df[df["Rating"] > 4.5]
print(highly_rating)
```

20.Find all free apps

```python
free_apps = df[df["Price"] == 0]
print(free_apps)
```

21.Filter apps with installs greater than 1,000,000

```python
print(df.loc[df["Installs"] > 1000000, ["App", "Installs"]])
```

22.Select apps belonging to multiple categories

```python
categories = ["GAME", "FAMILY", "TOOLS"]
print(df.loc[df["Category"].isin(categories), ["App", "Category"]])
```

23.Perform filtering using query syntax

```python
query_filtered = df.query("Rating > 4.0 and Price == 0")
print(query_filtered)
```

## 5. FEATURE ENGINEERING

24.Create a new column "Revenue" using installs and price

```python
df["Revenue"] = df["Installs"] * df["Price"]
print(df[["App", "Installs", "Price", "Revenue"]].head())
```

25.Categorize apps into rating groups (Low, Medium, High)

```python
df["Rating_Group"] = pd.cut(
    df["Rating"], bins=[0, 3.0, 4.2, 5.0], labels=["Low", "Medium", "High"]
)
print(df[["App", "Rating", "Rating_Group"]].head())
```

26.Create a new column to flag high-install apps (>1M installs)

```python
df["High Install"] = df["Installs"].apply(lambda x: 1 if x > 1000000 else 0)
print(df["High Install"])
```

27.Convert "Type" column into numeric values (Free=0, Paid=1)

```python
df["Type to Numeric"] = df["Type"].map({"Free": 0, "Paid": 1})
print(df["Type to Numeric"].unique())
```

### 6. SORTING & RANKING

28.Sort apps by number of installs in descending order

```python
df = df.sort_values(by="Installs", ascending=False)
print(df[["App", "Installs"]])
```

29.Identify top 10 apps based on number of reviews

```python
Top_10_apps = df.nlargest(10, "Reviews")
print(Top_10_apps[["App", "Reviews"]])
```

30.Identify bottom 10 apps based on installs

```python
bottom_10_installed = df.nsmallest(10, "Installs")
print(bottom_10_installed[["App", "Installs"]])
```

31.Rank apps based on rating

```python
df["Rating Rank"] = df["Rating"].rank(ascending=False, method="min")
print(df[["App", "Rating", "Rating Rank"]])
```

## 7. GROUPBY & AGGREGATION

32.Calculate average rating for each category

```python
avg_rating_per_cat = df.groupby("Category")["Rating"].mean()
print(avg_rating_per_cat)
```

33.Calculate total installs for each category

```python
total_installs_per_cat = df.groupby("Category")["Installs"].sum()
print(total_installs_per_cat)
```

34.Count the number of apps in each category

```python
app_count_per_cat = df["Category"].value_counts()
print(app_count_per_cat)
```

35.Perform multiple aggregations (mean rating, max installs) per category

```python
result = df.groupby("Category").agg({"Rating": "mean", "Installs": "max"})
print(result)
```

## 8. MERGE / JOIN / CONCAT

36.Split the dataset into multiple tables

```python
apps_info = df[["App", "Category", "Rating"]].drop_duplicates(subset=["App"])
installs_info = df[["App", "Installs", "Reviews"]].drop_duplicates(subset=["App"])
pricing_info = df[["App", "Type", "Price", "Revenue"]].drop_duplicates(subset=["App"])
print(apps_info.head())
print(installs_info.head())
print(pricing_info.head())
```

37.Merge app details with installs data using a common column

```python
merged_df = pd.merge(apps_info, installs_info, on="App")
print(merged_df.head())
```

38.Perform an inner join and observe the result

```python
inner_join = pd.merge(apps_info, installs_info, on="App", how="inner")
print(inner_join.head())
```

39.Perform a left join and identify missing values created after the merge

```python
left_merged = pd.merge(apps_info, pricing_info, on="App", how="left")
missing_after_left_join = left_merged.isnull().sum()
print(left_merged.head())
```
